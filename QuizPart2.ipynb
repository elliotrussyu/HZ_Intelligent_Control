*Question 5*

Iterations, tricky bastards

```python
a = 0
while a >= 0:
    print(a,end="\t")
    a += 1
    if a == 2:
        continue
    elif a == 5:
        break
    else:
        pass
```

```python
a = 0
while a >= 0:
    a += 1
    if a == 2:
        continue
    elif a == 5:
        break
    else:
        pass
    print(a,end="\t")
```

```python
letters = ["a","b","c","d"]
for i in range(len(letters)):
    print(letters[i],end='')
print("\n"+'-'*30)
print("".join([i for i in letters]))
print('-'*30)
for i,l in enumerate(letters):
    print(l+str(i),end=" ")
```
